**Lernziel:** Grundlagen der Datenanalyse - Mittelwerte, Gruppierung und einfache Visualisierung

**Vereinfachtes Szenario:** Analyse von Kundenalter und Ausgaben in verschiedenen Produktkategorien

**Lehrbeispiel!** Die Daten sind künstlich erzeugt. Alter, Produktkategorie und Ausgaben werden unabhängig voneinander simuliert; Unterschiede in der Stichprobe sind trotzdem möglich. Kein Bezug zu einem realen Unternehmen.

Wir betrachten nur Erwachsene ab 18 Jahren. **Kids** ist die gekaufte Produktkategorie, nicht das Alter der kaufenden Person (zum Beispiel kaufen Eltern Kinderkleidung). Die Ausgaben sind positiv und rechtsschief simuliert; das ist keine Behauptung über reale Kundendaten.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Einfache Simulationsdaten erstellen
np.random.seed(42)
n_customers = 200

data = {
    'alter': np.random.normal(35, 12, n_customers).round().astype(int),
    'kategorie': np.random.choice(['Damen', 'Herren', 'Kids'], n_customers),
    'ausgaben_euro': np.random.gamma(shape=8, scale=10, size=n_customers)
}

df = pd.DataFrame(data)
df = df.loc[df['alter'] >= 18].copy()
df['ausgaben_euro'] = df['ausgaben_euro'].round(2)

print(f"Anzahl Kunden: {len(df)}")
print(f"Ausgeschlossen, weil unter 18: {n_customers - len(df)}")
print("\nErste 5 Zeilen:")
df.head()
```

```python
# Einfache Deskriptive Statistiken
print("=== KUNDENANALYTIK EINES MODEHÄNDLERS ===")
print(f"Durchschnittsalter: {df['alter'].mean():.1f} Jahre")
print(f"Durchschnittliche Ausgaben: {df['ausgaben_euro'].mean():.2f} €")

print("\n=== AUSGABEN NACH KATEGORIE ===")
kategorie_stats = df.groupby('kategorie')['ausgaben_euro'].agg(['mean', 'median', 'std', 'count'])
print(kategorie_stats.round(2))
```

```python
# Einfache Visualisierung
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Balkendiagramm: Durchschnittliche Ausgaben pro Kategorie
kategorie_mean = df.groupby('kategorie')['ausgaben_euro'].mean()
ax1.bar(kategorie_mean.index, kategorie_mean.values, color=['skyblue', 'lightcoral', 'lightgreen'])
ax1.set_title('Durchschnittliche Ausgaben pro Kategorie')
ax1.set_ylabel('Ausgaben (€)')

# Streudiagramm: Alter vs. Ausgaben
ax2.scatter(df['alter'], df['ausgaben_euro'], alpha=0.6, color='darkblue')
ax2.set_title('Alter vs. Ausgaben')
ax2.set_xlabel('Alter (Jahre)')
ax2.set_ylabel('Ausgaben (€)')

plt.tight_layout()
plt.show()
```

## Zusammenfassung

**Was haben wir gelernt?**
- Daten simulieren, auf Erwachsene begrenzen und erste Übersicht verschaffen
- Mittelwerte und Gruppierungen berechnen  
- Einfache Visualisierungen erstellen
- Die Simulation enthält keine modellierten Zusammenhänge zwischen den Spalten. Trotzdem unterscheiden sich Gruppenmittelwerte zufällig; sie sind kein Beleg für echte Kategorie-Effekte.
- Aus diesen Grafiken allein lässt sich weder statistische Signifikanz noch das Fehlen eines Zusammenhangs ableiten. Hier wurde kein Hypothesentest durchgeführt.
- Eine angemessene Aussage lautet: **In dieser Stichprobe zeigt die deskriptive Darstellung keinen klaren Zusammenhang zwischen Alter und Ausgaben.** Das ist kein Nachweis von Unabhängigkeit in realen Daten.

**Einordnung für Vorlesung 10:** Dies ist eine deskriptive Vorübung, noch kein Clustering-Modell. Im Clustering-Notebook werden anschließend Ähnlichkeiten anhand von Merkmalen modelliert. Für echte Geschäftsentscheidungen wären unter anderem Stichprobenauswahl, Effektgrößen und Unsicherheit zu prüfen.
